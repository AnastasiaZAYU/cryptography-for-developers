## 🤖 Automated Reliability Tests

```python
import hashlib
from my_functions import *

def run_automated_tests():
  print("--- Starting Automated Tests ---")

  # 1. Parameter and Key Generation
  p_test, g_test = generate_parameters(512)
  a_test, b_test = generate_key_pair(p_test, g_test)
  print(f"[OK] Parameters and Keys generated.")

  # 2. Test Digital Signature
  msg_sig = "Verification of digital signature"
  r, s = sign(msg_sig, p_test, g_test, a_test)

  # Valid signature
  assert verify(msg_sig, r, s, b_test, p_test, g_test) == True, "Signature should be valid"
  # Invalid signature (modified message)
  assert verify(msg_sig + "!", r, s, b_test, p_test, g_test) == False, "Signature should be invalid for modified message"
  print(f"[OK] Digital Signature verified (Valid and Invalid cases).")

  # 3. Test Encryption/Decryption (Long message)
  msg_enc = "This is a long message that will be split into several blocks to test El-Gamal encryption logic."
  x, y = encrypt(msg_enc, b_test, p_test, g_test)
  decrypted_msg = decrypt(x, y, a_test, p_test)

  assert msg_enc == decrypted_msg, f"Decrypted message does not match original! Got: {decrypted_msg}"
  print(f"[OK] Encryption/Decryption verified for multi-block message.")

  print("\n--- All tests passed successfully! ---")

if __name__ == '__main__':
    run_automated_tests()
```
